# GeoLens — Quick Start

This notebook walks through the core GeoLens workflows:
1. Loading and inspecting a GeoTIFF
2. Tiling a large satellite scene for ML
3. Processing NISAR SAR data
4. Running a segmentation model


```python
import sys
sys.path.insert(0, '..')
from src import IsroGeoTiff, Nisar
from models import ResUNet, DeepLabV3, build_pspnet_v1, build_pspnet_v2
```

## 1. Load a GeoTIFF and inspect metadata

```python
tif = IsroGeoTiff(file_path='path/to/your_scene.tif')
metadata = tif.read_metadata()
print(metadata)
```

## 2. Extract a pixel-level subset

```python
subset = tif.raw_subset_from_tif(x=0, y=0, w=512, h=512)
subset.visualize_subset()
```

## 3. Tile a large scene for ML

```python
tiles = tif.tile_image(
    file_path='path/to/large_scene.tif',
    tile_width=256,
    tile_height=256,
    output_directory='./tiles'
)
print(f'Generated {len(tiles)} tiles')
```

## 4. Instantiate and compare segmentation models

```python
INPUT_SHAPE = (256, 256, 3)
NUM_CLASSES = 8

resunet   = ResUNet(INPUT_SHAPE, NUM_CLASSES)
deeplabv3 = DeepLabV3(INPUT_SHAPE, NUM_CLASSES)
pspnet_v1 = build_pspnet_v1(INPUT_SHAPE, NUM_CLASSES)
pspnet_v2 = build_pspnet_v2(INPUT_SHAPE, NUM_CLASSES)

for name, model in [('ResUNet', resunet), ('DeepLabV3+', deeplabv3),
                    ('PSPNet v1', pspnet_v1), ('PSPNet v2', pspnet_v2)]:
    params = model.count_params()
    print(f'{name:12s} → {params:,} parameters')
```
